# Histogram

```python
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
s = sp.io.mmread("mcmc_sample.mtx")
s = s.flatten()
s2 = sp.io.mmread("mcmc_sample2.mtx")
s2 = s2.flatten()
```

```python
fig, ax = plt.subplots(1, 2)
ax[0].hist(s, color = "black")
ax[1].hist(s2, color = "black")
plt.show()
```
