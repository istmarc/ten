# Histogram

```python
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.io.mmread("vect_norm.mtx")
x = x.flatten()
h = sp.io.mmread("hist.mtx")
h = h.flatten()
bins = sp.io.mmread("bins.mtx")
bins = bins.flatten()
```

```python
fig, ax = plt.subplots(1, 2)
ax[0].hist(x, color = "black")
ax[1].stairs(h, bins, color = "black", fill = "black")
plt.show()
```

## Cross validated histogram

```python
cvh = sp.io.mmread("cvhist.mtx")
cvh = cvh.flatten()
cvbins = sp.io.mmread("cvbins.mtx")
cvbins = cvbins.flatten()
```

```python
fig, ax = plt.subplots(1, 2)
ax[0].hist(x, color = "black")
ax[1].stairs(cvh, cvbins, color = "black", fill = "black")
plt.show()
```
